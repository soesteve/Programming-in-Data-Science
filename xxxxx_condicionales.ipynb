# Instrucciones condicionales


Las instrucciones condicionales permiten elegir una entre dos vías para resolver un problema, según sea el resultado de evaluar un predicado.

    Si predicado_se_cumple:
        hacer esto
    si no:
        hacer lo otro

Por ejemplo, para saber si alguien aprueba o no, basta con ver si la nota es un cinco, al menos:

    Si "la nota es un cinco al menos":
        la respuesta es que sí, que el estudiante aprueba
    si no:
        la respuesta es negativa, pues el estudiante no aprueba

La respuesta a la pregunta de si "la nota es al menos un cinco" es un sí o un no, esto es, True o False, que son los valores booleanos. En la instrucción condicional, siempre hay una expresión booleana que decide cuál es la instrucción que se realiza.

## Instrucción condicional completa

Lo anterior se escribe en Python así:

```python
def aprobado(nota: float)->bool:
    if nota >= 5.0:
        res = True
    else:
        res = False
    return res

print(aprobado(3.0), aprobado(7.5), aprobado(5.0), aprobado(10.0), aprobado(4.999))
```

Otra forma más compacta de definir la función aprobado es:

```python
def aprobado(nota: float)->bool:
    res = nota >= 5.0
    return res
print(aprobado(3.0), aprobado(7.5), aprobado(5.0), aprobado(10.0), aprobado(4.999))
```

Y todavía más compacta:

```python
def aprobado(nota: float)->bool:
    return nota >= 5.0
print(aprobado(3.0), aprobado(7.5), aprobado(5.0), aprobado(10.0), aprobado(4.999))
```

La función anterior está pobremente documentada, y queremos progresar tan rápido que a menudo omitimos la documentación. Hela aquí de nuevo con una documentación más completa:

```python
def aprobado(nota: float)->bool:
    """
    Averigua si una nota numérica es suficiente para aprobar o no


    Returns
    -------
    bool
        True, si la nota es suficiente para aprobar
        False, si la nota no lo es
    """
    if nota >= 5.0:
        res = True
    else:
        res = False
    return res

print(aprobado(3.0), aprobado(7.5), aprobado(5.0), aprobado(10.0), aprobado(4.999))
```

## Instrucción condicional incompleta

La rama "else" no siempre es necesaria:

```python
def valor_absoluto(x):
    if x < 0:
        x = -x
    return x

print(valor_absoluto(10), valor_absoluto(-10))
```

Proponer una documentación de la función anterior.

Otro ejemplo:

```python
def max_min(x, y):
    if x < y :
        max = y
        min = x
    else:
        max = x
        min = y
    return max, min   # Fíjate que se devuelven dos valores

print(max_min(2, 3), max_min(3, 2))
```

```python
a = 7
b = -1
x, y = max_min(a, b)       # asignación múltiple
print((x,y), max_min(-16, -8))
```

## Podemos anidar condicionales

```python
import math

def grado_poli(a:float, b:float, c:float) -> int:
    """
    Devuelve el grado del polinomio a*x^2 + b*x + c
    """
    if a == 0.0:
        if b == 0.0:
            grado = 0
        else:
            grado = 1
    else:
        grado = 2
    return grado

print(grado_poli(2, 3, 4), grado_poli(0, 1, 2), grado_poli(1, 0, 0), grado_poli(0, 0, 1))
```

## Elecciones múltiples

```python
def calificacion(num_nota: float)->str:
    """
    Converierte una calificación numérica en su denominación
    """
    if num_nota >= 9:
        nota = 'Sobresaliente'
    elif num_nota >= 7:
        # print(f'control 1: {num_nota}')
        nota = 'Notable'
    elif num_nota >= 5:
        nota = 'Aprobado'
    else:
        nota = 'Suspenso'
    return nota

print(calificacion(10), calificacion(1.23), calificacion(3.45), calificacion(4.999), calificacion(7), calificacion(5))
```

```python
n = 7.6
s = calificacion(n)
print('La nota alfabética de ', n,' es ',s)
```

```python
def bisiesto(a):
  if ((a % 4) != 0):
    resultado = "NO es bisiesto"
  elif ((a % 4) == 0) and ((a % 100) != 0):
    resultado =  " es bisiesto"
  elif ((a % 4) == 0) and ((a % 100) == 0) and ((a % 400) != 0):
    resultado =  " NO es bisiesto"
  elif ((a % 4) == 0) and ((a % 100) == 0) and ((a % 400) == 0):
    resultado =  " es bisiesto"
  return (a, resultado)
```
