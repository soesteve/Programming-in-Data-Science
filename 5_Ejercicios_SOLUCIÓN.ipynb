## 1. Años bisiestos

> En este ejercicio vamos a practicar if - elif. Para ello usaremos el operador `%` o __módulo__ (resto de la división) que devuelve el resto que queda al dividir un número entre otro.
> Por ejemplo, 10 % 3 es 1 porque al dividir 10 entre 3 obtenemos de resto 1.

   Un año es bisiesto si es múltiplo de 4, salvo si es múltiplo de 100 en cuyo caso debe ser múltiplo de 400.

Por ejemplo, Los años bisiestos entre 1998 y 2401 son:, __2000__, 2004, 2008, 2012, 2016, 2020, 2024, 2028, 2032, 2036, 2040, 2044, 2048, 2052, 2056, 2060, 2064, 2068, 2072, 2076, 2080, 2084, 2088, 2092, __2096, 2104__, 2108, 2112, 2116, 2120, 2124, 2128, 2132, 2136, 2140, 2144, 2148, 2152, 2156, 2160, 2164, 2168, 2172, 2176, 2180, 2184, 2188, 2192, __2196, 2204__, 2208, 2212, 2216, 2220, 2224, 2228, 2232, 2236, 2240, 2244, 2248, 2252, 2256, 2260, 2264, 2268, 2272, 2276, 2280, 2284, 2288, 2292, __2296, 2304__, 2308, 2312, 2316, 2320, 2324, 2328, 2332, 2336, 2340, 2344, 2348, 2352, 2356, 2360, 2364, 2368, 2372, 2376, 2380, 2384, 2388, 2392, 2396, __2400__

Implementa el cuerpo de la siguiente función.

```python
def es_bisiesto(a: int) -> bool:
    if a % 4 != 0:
        # Años que NO son divisibles por 4
        resultado = False

    elif a % 100 != 0:   
        # Años divisibles por 4 pero NO por 100
        # Ejemplos: 2004, 2008, 2012
        resultado = True

    elif a % 400 == 0:    
        # Años divisibles por 4, por 100 y por 400
        # Ejemplos: 2000, 2400
        resultado = True

    else:   
        # Años divisibles por 4 y por 100, pero NO por 400
        # Ejemplos: 2100, 2200, 2300, 2500
        resultado = False

    return resultado
```

## 2. Tabla de mutiplicar

> Este ejercicio debes hacerlo con un bucle ¿while o for?

Implementa una función que desarrolle la tabla de multiplicar de un número dado como entrada.

```python
def tabla_mult_for(n: int):
    print('\nTabla de multiplicar (for) del número: ', n, ' \n')   # Secuencia de escape \n es salto de línea 
    for x in range(1,11):
        print(n , '*', x, ' = ', n * x)
        x = x + 1
    print('---\n')

tabla_mult_for(2)
tabla_mult_for(5)

def tabla_mult_while(n: int):
    print('\nTabla de multiplicar (while) del número: ', n, ' n\n')   # Secuencia de escape \n es salto de línea 
    resultado = 0
    x = 1
    while (x <= 10):
        resultado = n + resultado
        print(n , '*', x, ' = ', resultado)
        x = x + 1
    print('---\n')

tabla_mult_while(2)
tabla_mult_while(5)
```

## 3. Sumar los primeros números enteros hasta uno dado


> Este ejercicio debes hacerlo con un bucle ¿while o for?

Deseamos sumar los primeros números enteros hasta uno dado. Por ejemplo, si el número es 10, la suma sería 1 + 2 + ... + 10. Por lo tanto, el resultado que devuelve la función es 55.


```python
def sumar_hasta_for(lim_sup: int) -> int:
    """
    Esta función calcula la suma de los enteros desde 1 hasta lim_sup,
    incluyendo ambos. Si lim_sup < 1, la función devuelve 0
    Returns
    -------
        Suma de 1 + 2 + 3 + ... + lim_sup

    Example
    -------
    >>> suma_hasta(4)
    10
    """

    print('\nSuma de e los enteros desde 1 hasta ', lim_sup, ' con for')  

    suma = 0

    for i in range(1, lim_sup + 1):
        suma = suma + i

    return suma

def sumar_hasta_while(lim_sup: int) ->int:
    """
    Esta función calcula la suma de los enteros desde 1 hasta lim_sup,
    incluyendo ambos. Si lim_sup < 1, la función devuelve 0
    Returns
    -------
        Suma de 1 + 2 + 3 + ... lim_sup

    Example
    -------
    >>> suma_hasta(4)
    10
    """

    print('\nSuma de e los enteros desde 1 hasta ', lim_sup, ' con while')  

    suma = 0
    i = 1
    while i <= lim_sup:
        suma = suma + i
        i = i + 1
    return suma
```

```python
sumar_hasta_for(100)
```

```python
sumar_hasta_while(100)
```

```python
sumar_hasta_for(-3)
```

```python
sumar_hasta_while(-3)
```

## Lista de los primeros `n` años bisiestos.

Escribe una función que devuelva en una lista los primeros `n` años bisiestos desde `1900`.

> En este ejercicio debes llamar a la función 'es_bisiesto' que has definido antes


```python
def primeros_bisiestos(n: int) -> list:
    """Devuelve los n primeros años bisiestos desde 1900."""
    resultado = []
    a = 1900

    print('Lista de los ', n, 'primeros años bisiestos desde 1900')
    while len(resultado) < n:
        if es_bisiesto(a):
            resultado.append(a)
        a += 1

    return resultado
```

```python
print(primeros_bisiestos(15))
```

## Buscar el siguiente año bisiesto

Un ejemplo de uso del while de forma natural es buscar el siguiente año bisiesto a un año dado. No sabemos cuántos años habrá que avanzar, puede ser un año, o dos, o lo que sea. Así que la estructura más adecuada es un `while`.

```python
def siguiente_bisiesto(a: int) -> int:
    """
    Devuelve el siguiente año bisiesto a partir del año indicado.
    """
    a += 1

    while not es_bisiesto(a):
        a += 1

    return a

print(siguiente_bisiesto(1999))     # 2000
print(siguiente_bisiesto(2000))     # 2004
print(siguiente_bisiesto(2003))     # 2004
print(siguiente_bisiesto(2096))     # 2104
print(siguiente_bisiesto(2398))     # 2400
```

### ¿Este ejercicio se puede hacer con un `for`?.

***Sí, se puede hacer con for, aunque, igual que en el ejercicio anterior, el while es más natural porque no sabemos cuántos años tendremos que comprobar hasta encontrar el siguiente bisiesto.***

***Es decir, conceptualmente no es una solución tan limpia.***
