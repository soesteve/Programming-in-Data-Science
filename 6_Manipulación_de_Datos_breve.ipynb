# Exploración y manipulación de Datos con Pandas

La exploración y manipulación de datos consiste en leer, organizar y modificar información almacenada en archivos o bases de datos. 
Para organizar nuestro trabajo, vamos a tener una carpeta llamada `Data` con los ficheros de datos de trabajo.

Python utiliza principalmente la biblioteca **Pandas** para la exploración y manipulación de datos y se importa mediante:

```python
import pandas as pd
```

donde `pd` es un alias, es decir, un nombre abreviado que nosotros elegimos para referirnos a la librería pandas.

La estructura principal de Pandas es el **DataFrame**, una tabla con filas y columnas similar a una hoja de cálculo (Excel, Number, Google sheet,...).
Podemos crear un DataFrame manualmente o importarlo desde un fichero.

## Crear un DataFrame manualmente:

````python
import pandas as pd

datos = {
    "Nombre": ["Ana", "Luis", "Marta"],
    "Edad": [24,31,28],
    "Ciudad": ["Madrid","Sevilla","Valencia"]
}

df = pd.DataFrame(datos)

df
````

## Seleccionar filas y columnas

También podemos seleccionar unas filas concretas con `loc`y `iloc`

- .loc: Selecciona columnas usando sus **nombres** o **etiquetas**.
- .iloc: Selecciona filas y/o columnas usando su **posición numérica** (enteros del 0 en adelante).

Nota: Los dos puntos ':' significa que trabajamos en un rango. Por ejemplo, `0:2` toma todos los números comprendidos entre `0` y `2` y tener los dos puntos sin números definiendo el rango `:`quiere decir que toma el rango completo.

````python
df.loc[3, "Nombre"]   # Celda
````

````python
df.loc[3, ["Nombre","Edad"]]   # fila 3 y columnas "Nombre","Edad"
````

````python
df.loc[3]      # fila 3  (todas las columnas)
````

````python
df.loc[:, ["Nombre","Edad"]]    # Todas las filas y columnas "Nombre","Edad"
````

````python
df.iloc[3,0]  # Celda
````

````python
df.iloc[3, [0,1]]      # fila 3 y columnas "Nombre","Edad"
````

````python
df.iloc[3]
````

````python
df.iloc[0:2]     # Filas 0:2 y todas las columnas
````

````python
df.iloc[:, [0, 1]]
````

## Agrupar

En un DataFrame, `groupby` sirve para agrupar las filas que tienen un mismo valor en una columna y después realizar alguna operación sobre cada grupo. Por ejmplo, podemos agrupar por población. Es decir, `groupby` divide el DataFrame en grupos según los valores de una o varias columnas y permite aplicar una operación a cada grupo.

```python
# Sintáxis general
df.groupby("columna")["columna_a_analizar"].operación()
```

```python
# Ejemplo
df.groupby("Población")
```

Esto crea los grupos, pero todavía no vemos un resultado útil. Tenemos que decir qué queremos hacer con cada grupo. 
Por ejemplo, queremos calcular la media edad.

````python
df.groupby("Población")["Edad"].mean()
````

````python
df
````

Podemos calcular varias operaciones

````python
df.groupby("Población")["Edad"].agg(["count", "mean", "min", "max"])
````

También podemos contar personas. Aquí debemos seleccionar otra columna como `Nombre`

````python
df.groupby("Población")["Nombre"].count()
````
