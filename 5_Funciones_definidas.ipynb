# Funciones definidas

<hr>

Una __función__ es un bloque de código **reutilizable** que realiza una tarea específica. Permite evitar repetir código, organizar mejor los programas y hacerlos más fáciles de mantener.

## Definición y llamada de una función

Una función se define con la palabra reservada `def`.

```python
def nombre_funcion(parametros):
    instrucciones
```

Un ejemplo de definición de función es:


````python
def calificar(num_nota: float)->str:
    if num_nota >= 9:
        nota = 'Sobresaliente'
    elif num_nota >= 7:
        # print(f'control 1: {num_nota}')
        nota = 'Notable'
    elif num_nota >= 5:
        nota = 'Aprobado'
    else:
        nota = 'Suspenso'
    return nota
````

### ¿Qué hace esta función?

La función recibe una nota numérica (por ejemplo, 8.5) y devuelve el texto correspondiente a esa nota (Notable). Habitualmente se dice que `8.5` es la entrada de la función `calificar` y `Notable` es la salida.

En concreto un función tiene las siguientes partes:

- Nombre de la función: __calificar__
- Parámetros de entrada: __num_nota__
- Cuerpo de la función: Todo lo que está indentado debajo de la cabeera de la función.
- Valor a devolver: __return__


Con respecto a la entrada, se puede componer de cero o varios parámetros y más concretamente:

- `num_nota` es el nombre de la variable que recibirá el dato.

- `float` indica que esperamos un número decimal.

De esta forma si hacemos la llamada a la función `calificar(8.5)` decimos que la variable `num_nota` toma el valor `8.5`.

El cuerpo de la función se compone de instrucciones, que trabajan con los parámetros.
Además hay una instrucción especial, llamada *return* sirve para expresar el valor que la función devuelve, tras los cálculos. En concreto, nuestra función devuelve una cadena de caracteres como indica `->str` que por cierto el tipado de la función (`float` y `->str`) es opcional. En Python, la anotación de tipo (type hint) no es obligatoria, pero sirve como ayuda para leer el código y para herramientas de programación.

El cuerpo de la función está indentado por unos espacios en blanco que están en el lado izquierdo de cada línea de código. En Python se usa la indentación para definir bloques de código. Las líneas que estén indentadas al mismo nivel pertenecen al mismo bloque de código.

Cada llamada de la función se hace con unos parámetros concretos, es decir, con los datos que realmente queremos hacer los cálculos. En cada llamada, los parámetros de entrada asumen los valores concretos y se ejecuta el cuerpo dichos datos. Por ejemplo, podemos hacer las siguientes múltiples llamadas.

````python
calificar(7)
````

````python
calificar(5.6)
````

````python
calificar(10)
````

## Documentación de una función (docstring)

Debemos indicar lo que hace la función y el tipo que deben tener los parámetros de entrada y el valor devuelto.

````python
def calificar(num_nota: float)->str:
    """
    Converierte una calificación numérica en su denominación
    """
    if num_nota >= 9:
        nota = 'Sobresaliente'
    elif num_nota >= 7:
        # print(f'control 1: {num_nota}')
        nota = 'Notable'
    elif num_nota >= 5:
        nota = 'Aprobado'
    else:
        nota = 'Suspenso'
    return nota
````
