## Problema: Los códigos como 0703 no deberían tratarse como números.

***Si Pandas intenta interpretar el código 0703 como un número, 0703 se convierte en 703, perdiendo el 0 inicial en la carga.***


### Solución

- Categóricas: todas las variables `cod_` y `des_`.
- Numéricas: `lat_`, `lon_`, nota_admision y variables de año.
- Indicadores: las variables `ind_` tienen como valores `SÍ` y `NO`. Las estudio por si puede pasarlas a booleanas.
- IMPORTANTE: variable especial que habría que revisar según su contenido.

Cambiar explícitamente el tipo de cada columna que puede generar el conflicto (Por ejemplo: `dtype={"IMPORTANTE": "str"}`) es la solución más adecuada, ya que nos permite controlar cómo interpreta Pandas los datos al leer el archivo.

En nuestro caso podemos cambiar `IMPORTANTE` por `str` o `category`. Si es `str`entonces se tratan a los valores como cadenas de caracters y si es `category` entonces, se tratan a los valores como un conjunto limitado de categorías, aunque sea muy, muy, muy grande, es limitado.

Pero son 78 variables. Se puede hacer ad-hoc o necesitamos un mecanismo para hacerla de forma automática y que identifique como:

- Categóricas: todas las variables `cod_` y `des_`.
- float: `lat_`, `lon_` y nota_admision
- int: 'anio_fin_estudio_acceso', 'anio_acceso_SUE', 'anio_nacimiento'.
- Booleanas: las variables `ind_` con valores `SÍ` y `NO`. 
- str: la variable IMPORTANTE.

Pero hay un problema importante con las variables `ind_` porque en el CSV tienen los valores `SÍ` y `NO`, hacer un `dtype = "boolean"` no los convierte utomáticamente a True y False. En este caso, es mejor leer primero las variables `ind_` como `string` y después convertirla.

Hay que hacerlo en dos pasos: 

1. Leer el CSV con los tipos que se pueden aplicar directamente
2. Convertir las ind_ de SÍ/NO a booleanas.



```python
import pandas as pd

fichero = "./Data/ucm-acceso-2024-25-anonimizado.csv"


# --------------------------------------------------
# 1. Obtener los nombres de las columnas
# --------------------------------------------------

columnas = pd.read_csv(fichero, nrows=0).columns


# --------------------------------------------------
# 2. Identificar las columnas categóricas
# --------------------------------------------------

columnas_cat = []

for c in columnas:
    if c.startswith(("cod_", "des_")):
        columnas_cat.append(c)


# --------------------------------------------------
# 3. Identificar las columnas Float64
# --------------------------------------------------

columnas_float = []

for c in columnas:
    if c.startswith(("lat_", "lon_")) or c == "nota_admision":
        columnas_float.append(c)


# --------------------------------------------------
# 4. Variables enteras
# --------------------------------------------------

columnas_int = [
    "anio_fin_estudio_acceso",
    "anio_acceso_SUE",
    "anio_nacimiento"
]


# --------------------------------------------------
# 5. Identificar las variables booleanas
# --------------------------------------------------

columnas_bool = []

for c in columnas:
    if c.startswith("ind_"):
        columnas_bool.append(c)


# --------------------------------------------------
# 6. Definir los tipos de datos para la carga
# --------------------------------------------------

dtype = {}

# Las categorías las cargamos primero como string.
# Después las convertiremos a category.
for c in columnas_cat:
    dtype[c] = "string"    # ¿categórica?

for c in columnas_float:
    dtype[c] = "string"

# Los años se cargan como string porque pueden aparecer
# valores como 2008.
for c in columnas_int:
    dtype[c] = "string"

# SÍ / NO se cargan como string y después se convierten a boolean.
for c in columnas_bool:
    dtype[c] = "string"

dtype["IMPORTANTE"] = "string"


# --------------------------------------------------
# 7. Leer el fichero
# --------------------------------------------------

df = pd.read_csv(
    fichero,
    dtype=dtype
)

# --------------------------------------------------
# 8. Convertir las columnas float
# --------------------------------------------------

for c in columnas_float:
    df[c] = (
        df[c]
        .str.strip()
        .str.replace(",", ".", regex=False)
        .astype("Float64")
    )
    
# --------------------------------------------------
# 8. Convertir los códigos a category
# --------------------------------------------------

for c in columnas_cat:
    df[c] = df[c].astype("category")


# --------------------------------------------------
# 9. Convertir los años a enteros
# --------------------------------------------------

for c in columnas_int:
    df[c] = (
        df[c]
        .str.strip()
        .str.replace(".", "", regex=False)
        .astype("Int64")
    )


# --------------------------------------------------
# 10. Convertir SÍ / NO a boolean
# --------------------------------------------------

for c in columnas_bool:
    df[c] = (
        df[c]
        .str.strip()
        .map({
            "SÍ": True,
            "SI": True,
            "NO": False
        })
        .astype("boolean")
    )


# --------------------------------------------------
# 11. Comprobar el resultado
# --------------------------------------------------

df.info()
```
