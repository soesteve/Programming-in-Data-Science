## Filtrar

Se **filtran** las filas u observaciones que cumplen una determinada restricción.

```python
# df[df["nota_admision"] >= 10] da error de tipo
# Ese error suele aparecer porque la columna nota_admision está almacenada como texto (object o string) y no como números. 
# Debes convertir la columna a numérica, para ello usa
# pd.to_numeric() convierte la columna a un tipo numérico.
# errors="coerce" convierte los valores no válidos en NaN.
```

```python
df["nota_admision"] = pd.to_numeric(df["nota_admision"], errors="coerce") 
```

```python
df[df["nota_admision"] >= 10]
```

```python
df[(df["nota_admision"] >= 10) & (df["des_centro"] == "Facultad de Derecho")]
```

## Añadir una fila con un valor nulo

Para añadir una fila, decimos que pasada la última fila (len(df)) se edita df.loc[ ] y se le asigna una lista de datos. En este caso, voy a aprovechar que añado una fila para dejar una celda sin dato. Por ejemplo, dejamos la edad sin conocer.

Aviso: cada vez que ejecutes esta intrucción, se añade una fila nueva.

```python
df.loc[len(df)] = ["Amelia", None, "Toledo"]
```
