# Control de flujo

El tema de Control de flujo explica cómo un programa decide qué instrucciones ejecutar y cuántas veces repetirlas. Es decir, el control de flujo determina el orden en que se ejecutan las instrucciones de un programa. Por defecto, las instrucciones se ejecutan una detrás de otra (ejecución secuencial). Sin embargo, muchas veces necesitamos:

* Tomar decisiones ("si ocurre esto, haz una cosa; si no, haz otra cosa").

* Repetir acciones varias veces.

Python, como todos los lenguajes de programación, proporciona varias herramientas para controlar el flujo del programa.

## La instrucción `if`

La instrucción `if` permite ejecutar un bloque de código únicamente cuando una condición es verdadera.
Su sintáxis es:

>```
>if condicion:
>    instrucciones
>```

Observa que:
* La condición termina con :

* El bloque que se ejecuta está indentado (normalmente con cuatro espacios)

* Si la condición es falsa, no ocurre nada

````python
# Ejemplo
edad = 28

if edad >= 18:
    print("Eres mayor de edad.")
````

## La instrucción `if-else`

Cuando queremos elegir entre dos alternativas utilizamos `else`.

>```python
>if condicion:
>    instrucciones
>else:
>    instrucciones
>```


````python
# Ejemplo
edad = 15

if edad >= 18:
    print("Mayor de edad")
else:
    print("Menor de edad")
````

## La instrucción `if-elif-else`

Cuando existen varias posibilidades utilizamos `elif`.

````python
# Ejemplo

nota = 7

if nota >= 9:
    print("Sobresaliente")
elif nota >= 7:
    print("Notable")
elif nota >= 5:
    print("Aprobado")
else:
    print("Suspenso")
````

Las condiciones se evalúan de arriba hacia abajo y solo se ejecuta el primer bloque cuya condición sea verdadera.

## El bucle `for`

El bucle `for` repite un bloque de instrucciones recorriendo una secuencia. 

>```python
>for variable in iterable:
>    # Bloque de código que se repite
>```

````python
# Ejemplo
frutas = ["manzana", "pera", "kiwi"]

for fruta in frutas:
    print(fruta)
````

Podemos usar los datos contenidos en una lista, ejemplo anterior, o usar un rango numérico `range()` que genera una secuencia de números. Además, `range()` puede puede variar el número de sus parámetros.

* `range()` con 1 parámetro: muestra el rango desde 0 hasta dicho núnero.
```
    range(5) devuelve 0,1,2,3,4
```
* `range()` con 2 parámetros: muestra el rango que definen dichos parámetros
```
    range(2,6) devuelve 2,3,4,5
```
* `range()` con 3 parámetros: muestra el rango que definen los dos primeros parámetros con el salto que define el tercer parámetro.
```
    range(0,10,2) devuelve 0,2,4,6,8
```

````python
for i in range(5):
    print(i)
````

````python
for numero in range(2, 11, 2):
    print(numero)
````

## El bucle while

`while` repite instrucciones mientras una condición sea verdadera. Su sintaxis es:

>```python
>while condicion:
>    instrucciones
>```

````python
contador = 1

while contador <= 5:
    print(contador)
    contador += 1
````

Error común: NO actualizar la variable de control nos lleva a un bucle infinito.

## Comparación entre for y while

|for (recorrido)|while (búsqueda)|
|---|-----|
|Recorre una secuencia | Depende de una condición|
|Número de repeticiones conocido antes de empezar | Número de repeticiones desconocido|
