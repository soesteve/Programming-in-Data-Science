## Problema: Los códigos como 0703 no deberían tratarse como números.

***Si Pandas intenta interpretar el código 0703 como un número, 0703 se convierte en 703, perdiendo el 0 inicial.***


### Solución 1

Decir a Pandas que lea el archivo de otra manera para determinar los tipos de las columnas (`low_memory = False`).



```python
df = pd.read_csv(
    "./Data/ucm-acceso-2024-25-anonimizado.csv",
    low_memory = False   
)

df[['cod_especialidad_acceso']]
```

### Solución 2

Cambiar explícitamente el tipo de la columna que puede generar el conflicto (`dtype={"cod_especialidad_acceso": "category"}`) es la solución más adecuada, ya que nos permite controlar cómo interpreta Pandas los datos al leer el archivo.

En nuestro caso podemos cambiarla por `str` o `category`. Si es `str`entonces se tratan a los valores como cadenas de caracters y si es `category` entonces, se tratan a los valores como un conjunto limitado de categorías, aunque sea muy, muy, muy grande, es limitado.

```python
df = pd.read_csv(
    "./Data/ucm-acceso-2024-25-anonimizado.csv",
    dtype={"cod_especialidad_acceso": "category"}
)

df[['cod_especialidad_acceso']]
```

De hecho, creo que es conveniente mirar **todas** las columnas numéricas.

```python
df.describe()
```

Veo que ¿todas? las variable numéricas comienzan por `cod_` y deberían ser categóricas (NO str), si tengo dudas tengo que consultar la documentación:
https://dimetrical.atlassian.net/wiki/spaces/UNC/pages/1608286213/Dataset+Acceso

Cambio todos los tipos de las columnas que representan códigos a categóricas.

```python
for columna in df.columns:
    if columna.startswith("cod_"):
        df[columna] = df[columna].astype("category")

df.describe()
```

Pues no eran todas. Hay variables numéricas que denotan años, latitud y longitud. pero puedo quitar las descripciones `_des`.

```python
for columna in df.columns:
    if columna.startswith("des_"):
        df[columna] = df[columna].astype("category")
        
df.describe()
```
