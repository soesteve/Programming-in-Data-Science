## ¿Qué es la visualización de datos?

La visualización de datos consiste en representar información mediante gráficos para facilitar su comprensión.

Un gráfico puede ayudarnos a:

- **Comparar** cantidades entre categorías.
- **Observar la evolución** de una variable a lo largo del tiempo.
- **Estudiar la distribución** de los datos.
- **Analizar relaciones** entre dos variables.
- **Detectar patrones** o valores que se alejan del resto.

No existe un único gráfico adecuado para todos los problemas. La elección depende de la pregunta que queramos responder.

## Introducción guiada por un ejemplo

### Primer paso: preparar los datos

Vamos a utilizar un conjunto de datos pequeño sobre las ventas de una tienda durante varios meses.
Cada fila representa un mes y contiene el número de unidades vendidas y los ingresos obtenidos.

````python
import pandas as pd

datos = {
    "Mes": ["Enero", "Febrero", "Marzo", "Abril", "Mayo", "Junio"],
    "Unidades": [120, 150, 135, 180, 210, 195],
    "Ingresos": [2400, 3000, 2700, 3600, 4200, 3900]
}

df = pd.DataFrame(datos)
df
````

### Segundo paso: crear un gráfico 

Para el `df` que tenenmos, un *gráfico de líneas* es útil para observar cómo cambia una variable a lo largo del tiempo.
Por ejemplo, podemos estudiar la evolución de las unidades vendidas mes a mes.
Pero antes de crear un gráfico, necesitamos importar la biblioteca `matplotlib` y en concreto `pyplot` que es el módulo de `Matplotlib` que contiene funciones para crear y modificar gráficos. Además, creamos un alias con `plt`.

Vamos a crear un gráfico para explicar qué significa cada instrucción y por qué se utiliza.

````python
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 3))   
plt.plot(df["Mes"], df["Unidades"], marker="o")   
plt.title("Unidades vendidas por mes")
plt.xlabel("Mes")
plt.ylabel("Unidades vendidas")
plt.grid(True)
plt.show()
````

Toda instrucción que comienza por `plt`no son órdenes propias de Python básico, son funciones de la biblioteca `matplotlib`.

**Explicación de cada instrucción**

- `plt.figure(figsize=(6, 3))`: La función `plt.figure()` crea una nueva figura, es decir, el espacio donde se dibujará el gráfico y tiene el argumento `figsize`que se instancia (`=`) a un tamaño dado '(ancho, alto)'. No es obligatorio indicar el tamaño. Si no lo hacemos, Matplotlib utiliza un tamaño predeterminado.
- `plt.plot(df["Mes"], df["Unidades"], marker="o")`: Dibuja un gráfico de líneas.
La función `plot()` recibe en sus argumentos los datos que se quieren representar, `df["Mes"]`son valores del eje horizontal y `df["Unidades"]`son los valores del eje vertical. Además, se puede indicar `marker="o"`que dibuja un círculo en cada punto de datos.
- `plt.title("Unidades vendidas por mes")`: Añade el título del gráfico. 
- `plt.xlabel("Mes")`: Añade la etiqueta del eje horizontal, que representa los meses.
- `plt.ylabel("Unidades vendidas")`: Añade la etiqueta del eje vertical.
- `plt.grid(True)`: Muestra una cuadrícula de referencia que facilita la lectura de los valores del gráfico.


Como podemos observar, hay dos tipos diferentes de funciones:
- Las funciones para crear gráficos, como `plt.plot()`.
- Las funciones para configurar los gráficos, como `plt.figure()`, `plt.title()` o `plt.xlabel()`.

Empecemos por las funciones que crean gráficos

### Gráfico de líneas, `plt.plot()`: 


Sirve para observar la evolución de una variable, por ejemplo, las ventas a lo largo de los meses.
Las líneas son especialmente útiles cuando el orden de los datos tiene significado, por ejemplo, en una serie temporal.

````python
plt.figure(figsize=(4, 2))
plt.plot(df["Mes"], df["Unidades"], marker="o")
plt.title("Unidades vendidas por mes")
plt.xlabel("Mes")
plt.ylabel("Unidades vendidas")
plt.grid(True)
plt.show()
````

**Interpretación:** cada punto representa las ventas de un mes. Las líneas ayudan a ver tendencias y cambios.

**Posibles preguntas**

- ¿En qué mes se vendieron más unidades?
   
    En mayo, con 210 unidades vendidas.

- ¿En qué mes disminuyeron las ventas respecto al mes anterior?

   En marzo, cuando las ventas bajaron de 150 a 135 unidades, y en junio, cuando bajaron de 210 a 195 unidades.

- ¿Cuál es la tendencia general durante el periodo?

 La tendencia general es creciente: las ventas pasan de 120 unidades en enero a 195 en junio, aunque hay descensos en marzo y junio.


### Gráfico de barras: comparar categorías

Un gráfico de barras permite comparar cantidades entre distintas categorías. Por ejemplo, podemos comparar las ventas de cada mes.

````python
plt.figure(figsize=(6, 2))
plt.bar(df["Mes"], df["Unidades"])
plt.title("Comparación de unidades vendidas")
plt.xlabel("Mes")
plt.ylabel("Unidades vendidas")
plt.show()
````

**Interpretación:** la altura de cada barra indica las unidades vendidas.

**Posibles preguntas**

- ¿Qué categoría tiene mayor valor?

    Mayo, con 210 unidades.
 
- ¿Qué diferencia hay entre enero y mayo?

    \[210-120=90\]. En mayo se vendieron 90 unidades más que en enero.
  
- ¿Qué ventaja ofrece este gráfico frente al gráfico de líneas para comparar meses concretos?

  Las barras permiten comparar visualmente las cantidades de cada mes. El gráfico de líneas facilita observar la evolución y las tendencias entre meses consecutivos. Ambos pueden servir para comparar valores, pero cada uno enfatiza un aspecto diferente.

**Gráfico de barras horizontales `plt.barh()`**.

Es útil cuando las categorías tienen nombres largos

````python
productos = ["Ordenador portátil", "Monitor", "Teclado", "Ratón"]
unidades_producto = [35, 24, 50, 65]

plt.figure(figsize=(4, 2))
plt.barh(productos, unidades_producto)
plt.title("Unidades vendidas por producto")
plt.xlabel("Unidades vendidas")
plt.ylabel("Producto")
plt.show()
````

## Histograma: estudiar una distribución


Un histograma representa la distribución de una variable numérica. Agrupa valores numéricos en intervalos y muestra cuántas observaciones pertenecen a cada intervalo. 

**Importante:** A diferencia de un gráfico de barras, se utiliza para estudiar la distribución de una variable numérica. De hecho, en un histograma las barras representan intervalos consecutivos de una variable numérica. En un gráfico de barras, en cambio, se comparan categorías diferenciadas.

Por ejemplo, si tenemos las notas de 30 estudiantes, queremos saber si se concentran alrededor del aprobado, si predominan las notas altas o si hay muchas notas bajas.

````python
calificaciones = [
    4.2, 5.1, 6.3, 7.0, 5.8, 8.1, 6.7, 7.4, 9.0, 5.5,
    6.1, 7.8, 8.3, 4.9, 6.9, 7.2, 5.7, 8.7, 6.5, 7.1,
    3.8, 4.5, 9.2, 6.0, 7.6, 8.0, 5.3, 6.8, 7.9, 5.9
]

plt.figure(figsize=(4, 2))
plt.hist(calificaciones, bins=6, edgecolor="black")
plt.title("Distribución de las calificaciones")
plt.xlabel("Calificación")
plt.ylabel("Número de estudiantes")
plt.show()
````

¿Qué significa cada argumento?
- calificaciones: lista de valores numéricos que queremos representar.
- bins=6: divide el rango de los datos en seis intervalos.
    Así, los intervalos y sus frecuencias son:

    | Intervalo aproximado | Número de calificaciones |
    |---|---:|
    | 3,8–4,7 | 3 |
    | 4,7–5,6 | 4 |
    | 5,6–6,5 | 6 |
    | 6,5–7,4 | 7 |
    | 7,4–8,3 | 6 |
    | 8,3–9,2 | 4 |

- edgecolor="black": dibuja los bordes de las barras en negro.

El eje horizontal representa los intervalos de notas y el vertical indica cuántos estudiantes tienen notas dentro de cada intervalo.
Los extremos exactos dependen de cómo se definan los intervalos; el último incluye el valor máximo.

En general:
- Pocos intervalos: resumen más general de la distribución.
- Muchos intervalos: más detalle, aunque puede resultar más difícil identificar el patrón global.
No existe un número de intervalos óptimo para todos los conjuntos de datos.

**Interpretación**


**Posibles preguntas**

- ¿Cómo se distribuyen valores numéricos?

  Las calificaciones se concentran principalmente entre 5,6 y 8,3, con mayor presencia en los intervalos centrales y relativamente pocas notas muy bajas o muy altas
  
- ¿En qué intervalo se concentran más calificaciones?

  En el intervalo comprendido aproximadamente entre 6,5 y 7,4, con 7 calificaciones.
  
- ¿Hay muchas calificaciones muy bajas?

    No. Solo cuatro calificaciones son inferiores a 5: 3,8; 4,2; 4,5 y 4,9.
  
- ¿Qué información ofrece el histograma que no se ve tan fácilmente en una lista de números?

    Permite observar de un vistazo dónde se concentran las calificaciones, cuáles son los intervalos más frecuentes y si existen pocos o muchos valores en los extremos.

## Diagrama de dispersión `plt.scatter()`: relación entre variables

Un diagrama de dispersión permite explorar si existe una relación entre dos variables numéricas. Si los puntos muestran una tendencia, podemos observar una asociación entre las variables. Sin embargo, el gráfico por sí solo no demuestra que una variable cause la otra.


En este ejemplo, comparamos las horas de estudio con la calificación obtenida por varios estudiantes.

````python
horas_estudio = [1, 2, 2.5, 3, 3.5, 4, 4.5, 5, 6, 7]
notas =         [3.2, 4.1, 5.0, 5.4, 6.2, 6.8, 7.1, 7.9, 8.4, 9.3]

plt.figure(figsize=(7, 4))
plt.scatter(horas_estudio, notas)
plt.title("Horas de estudio y calificación")
plt.xlabel("Horas de estudio")
plt.ylabel("Calificación")
plt.grid(True)
plt.show()
````

**Interpretación**

En estos datos, las notas tienden a aumentar con las horas de estudio. Una asociación observada en un gráfico no demuestra por sí sola causalidad

**Posibles preguntas**

- ¿Parece existir una relación entre las horas de estudio y la calificación?

  Sí. En los datos del ejemplo, las calificaciones aumentan a medida que aumentan las horas de estudio. Se observa una asociación positiva.
  
- ¿La relación es perfecta?

  No. Aunque las calificaciones aumentan en todas las observaciones, los puntos no siguen una relación lineal perfecta: los incrementos de nota no son proporcionales a los incrementos de horas.
  
- ¿Podemos concluir que estudiar más horas causa siempre una nota más alta?

    No. El gráfico muestra una asociación, pero no demuestra causalidad. También pueden influir otros factores, como los conocimientos previos, el método de estudio o la dificultad del examen.

Un gráfico puede sugerir una asociación, pero por sí solo no demuestra que una variable sea la causa de la otra.

### Gráfico de sectores: `plt.pie()`

Representa cómo se reparte un total entre categorías. Es más fácil de interpretar cuando hay pocas categorías. Los argumentos de la función `plt.pie()`son `labels` que muestra las categorías y `autopct` que escribe el porcentaje de cada sector siguiendo el formato:

| Elemento | Significado |
|---|---|
| `%` | Inicia una especificación de formato. |
| `1` | Anchura mínima del campo. |
| `.1` | Indica que queremos un decimal. |
| `f` | Representa un número decimal. |
| `%%` | Muestra el símbolo `%` literal. |

````python
categorias = ["Informática", "Libros", "Accesorios", "Otros"]
ventas = [40, 25, 20, 15]
plt.figure(figsize=(3, 3))
plt.pie(ventas, labels=categorias, autopct="%1.1f%%")
plt.title("Distribución de las ventas por categoría")
plt.show()
````

### Diagrama de caja: plt.boxplot()

Un diagrama de caja, también llamado boxplot, resume una distribución numérica mediante cuartiles, mediana y posibles valores atípicos. Es útil cuando queremos conocer la posición central de los datos, su dispersión y si hay observaciones que se alejan del resto.

Los elementos del diagrama de caja

- Mediana (Q2)
Divide los datos ordenados en dos mitades: aproximadamente la mitad queda por debajo y la otra mitad por encima.

- Primer y tercer cuartil (Q1 y Q3)
Delimitan la caja. Q1 corresponde al percentil 25 y Q3 al percentil 75.

- Bigotes
Se extienden hasta los valores más extremos que no se consideran atípicos según la regla habitual de Matplotlib.

- Valores atípicos potenciales
Se representan individualmente cuando quedan fuera de los límites de los bigotes.

El rango intercuartílico se calcula como:
    \[RIC = Q3-Q1\]

Los límites habituales para identificar posibles valores atípicos son:
    \[L_{\mathrm{inf}}=Q1-1{,}5\,RIC\]

    \[L_{\mathrm{sup}}=Q3+1{,}5\,RIC\]

Los valores que quedan fuera de esos límites se representan como puntos separados. Esto no significa necesariamente que sean errores: podrían ser observaciones válidas.

````python
notas_grupo = [4.5, 5.0, 5.2, 5.5, 5.8, 6.0, 6.1, 6.3, 6.5, 6.7,
               6.8, 7.0, 7.1, 7.2, 7.5, 7.7, 8.0, 8.2, 8.5, 10.0]

# RRREEEPAAAASSSAAARRR

#print(notas_grupo.info())

plt.figure(figsize=(4, 4))
plt.boxplot(notas_grupo)
plt.title("Distribución de las notas")
plt.ylabel("Nota")
plt.show()
````

El diagrama de caja resume la distribución sin necesidad de mostrar cada observación individual.

- Mediana: 6,75.
- Primer cuartil: aproximadamente 5,95.
- Tercer cuartil: aproximadamente 7,55.
- Valor potencialmente atípico: 10,0, que queda ligeramente por encima del límite superior habitual de 1,5 veces el rango intercuartílico.


## Funciones de configuración

Las funciones de configuración nos permiten personalizar los gráficos modificando elementos como el tamaño, el título, las etiquetas, la leyenda, la cuadrícula o los límites de los ejes. Estas funciones se  resumen en la siguiente tabla. Además, también veremos cómo modificar colores, tipos de línea y marcadores mediante argumentos como `color`, `linestyle`, `linewidth` y `marker`.
En Matplotlib, algunas instrucciones habituales son:


| Función | Utilidad |
|---|---|
| `plt.figure()` | Crear una figura y configurar su tamaño. |
| `plt.title()` | Añadir un título. |
| `plt.xlabel()` | Etiquetar el eje X. |
| `plt.ylabel()` | Etiquetar el eje Y. |
| `plt.legend()` | Mostrar una leyenda. |
| `plt.grid()` | Activar o desactivar la cuadrícula. |
| `plt.xlim()` | Establecer los límites del eje X. |
| `plt.ylim()` | Establecer los límites del eje Y. |
| `plt.xticks()` |	Configurar las marcas y etiquetas del eje X. |
| `plt.yticks()` |	Configurar las marcas y etiquetas del eje Y. |
| `plt.savefig()` | Guardar el gráfico en un fichero. |


Para introducir `plt.legend()`, podemos representar dos series en el mismo gráfico y añadir una leyenda.
Las etiquetas `label` dan nombre a cada serie y `plt.legend()` muestra esos nombres. Observa que unidades e ingresos tienen escalas diferentes, por lo que compartir un eje vertical puede dificultar la comparación.

````python
plt.figure(figsize=(6, 3))

plt.plot(df["Mes"], df["Unidades"], marker="o", label="Unidades")
plt.plot(df["Mes"], df["Ingresos"], marker="s", label="Ingresos")

plt.title("Unidades e ingresos por mes")
plt.xlabel("Mes")
plt.ylabel("Valor")
plt.legend()
plt.grid(True)
plt.show()
````

**Pregunta para debatir:** ¿es adecuado representar unidades e ingresos en el mismo eje vertical? ¿Qué problema puede aparecer cuando las variables tienen escalas muy diferentes?
El código funciona, pero las unidades y los ingresos tienen escalas y unidades de medida diferentes. 
Las unidades se encuentran aproximadamente entre 120 y 210, mientras que los ingresos están entre 2.400 y 4.200 euros. Al compartir el mismo eje, la serie de unidades queda muy cerca de la parte inferior y resulta difícil interpretarla.
El gráfico puede inducir a interpretaciones erróneas. Es mejor representarlos en dos gráficos separados, cada uno con su propia escala y etiquetas.


Por otra parte podemos cambiar límites de los ejes: `plt.xlim()` y `plt.ylim()` con:

- `plt.xlim(minimo, maximo)` establece los límites visibles del eje X.
- `plt.ylim(minimo, maximo)` establece los límites visibles del eje Y.

Elige límites que ayuden a interpretar los datos. Es como una *lupa*.

````python
plt.figure(figsize=(6, 3))
plt.plot(df["Mes"], df["Unidades"], marker="o")
plt.title("Unidades vendidas por mes")
plt.ylim(0, 250)
plt.show()
````

Además, podemos cambiar la orientación de las etiquetas y guardar el gráfico.

````python
plt.figure(figsize=(8, 4))
plt.plot(df["Mes"], df["Unidades"], marker="o")
plt.title("Unidades vendidas por mes")
plt.xlabel("Mes")
plt.ylabel("Unidades vendidas")
plt.xticks(rotation=45)
plt.yticks(rotation=15)
plt.savefig("unidades_por_mes.png", dpi=150, bbox_inches="tight")
plt.show()
````
