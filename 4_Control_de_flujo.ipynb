# Control de flujo

El tema de Control de flujo explica cómo un programa decide qué instrucciones ejecutar y cuántas veces repetirlas. Es decir, el control de flujo determina el orden en que se ejecutan las instrucciones de un programa. Por defecto, las instrucciones se ejecutan una detrás de otra (ejecución secuencial). Sin embargo, muchas veces necesitamos:

* Tomar decisiones ("si ocurre esto, haz una cosa; si no, otra").

* Repetir acciones varias veces.

* Detener o continuar una repetición según una condición.

Python, como todos los lenguajes de programación, proporciona varias estructuras para controlar el flujo del programa.

## La estructura `if`

La instrucción `if` permite ejecutar un bloque de código únicamente cuando una condición es verdadera.
Su sintáxis es:

```
if condicion:
    instrucciones
```

Observa que:
* La condición termina con :.

* El bloque que se ejecuta está indentado (normalmente con cuatro espacios).

* Si la condición es falsa, no ocurre nada.

````python
# Ejemplo
edad = 28

if edad >= 18:
    print("Eres mayor de edad.")
````

## La estructura `if-else`

Cuando queremos elegir entre dos alternativas utilizamos `else`.

```
if condicion:
    instrucciones
else:
    instrucciones
```


````python
# Ejemplo
edad = 15

if edad >= 18:
    print("Mayor de edad")
else:
    print("Menor de edad")
````

## La estructura `if-elif-else`

Cuando existen varias posibilidades utilizamos `elif`.

````python
# Ejemplo

nota = 7

if nota >= 9:
    print("Sobresaliente")
elif nota >= 7:
    print("Notable")
elif nota >= 5:
    print("Aprobado")
else:
    print("Suspenso")
````

Las condiciones se evalúan de arriba hacia abajo y solo se ejecuta el primer bloque cuya condición sea verdadera.

## El bucle `for`

El bucle `for` repite un bloque de instrucciones recorriendo una secuencia. 

```
for variable in iterable:
    # Bloque de código que se repite
```

````python
# Ejemplo
frutas = ["manzana", "pera", "kiwi"]

for fruta in frutas:
    print(fruta)
````

Podemos usar los datos contenidos en una lista, ejemplo anterior, o usar un rango numérico. `range()` genera una secuencia de números, dependiendo del número de parámetros de la función `range()`:
* 1 parámetro: muestra el rango desde 0 hasta dicho núnero.
```
    range(5) devuelve 0,1,2,3,4
```
* 2 parámetros: muestra el rango que definen dichos parámetros
```
    range(2,6) devuelve 2,3,4,5
```
* 3 parámetros: muestra el rango que definen los dos primeros parámetros con el salto que define el tercer parámetro.
```
    range(0,10,2) devuelve 0,2,4,6,8
```

````python
for i in range(5):
    print(i)
````

````python
for numero in range(2, 11, 2):
    print(numero)
````

## El bucle while

`while` repite instrucciones mientras una condición sea verdadera. Su sintaxis es:
```
while condicion:
    instrucciones
```

````python
contador = 1

while contador <= 5:
    print(contador)
    contador += 1
````

Error común: NO actualizar la variable de control ==> un bucle infinito.

## Comparación entre for y while

|for|while|
|---|-----|
|Recorre una secuencia|Depende de una condición|
|Número de repeticiones conocido|Número de repeticiones desconocido|
