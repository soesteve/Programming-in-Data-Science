## Inscripciones a un taller de cocina

En un taller de cocina algunas personas se han apuntado más de una vez por error. 
    Utiliza una lista y un conjunto para ver la diferencia.


```python
inscripciones = ["Ana", "Carmen", "Jose", "Conchi", "Lucía", "Alejandra", "Sonia", "Luis", "Ana", "Marta", "Luis", "Sonia"]
```

Realiza las siguientes acciones:

1. Muestra la lista.

2. ¿Cuántas inscripciones hay?.

3. Añade a Carlos al final.

4. Vuelve a mostrar la lista y el número de inscripciones.

5. Ahora convierte la lista en un conjunto. Pista: set(inscripciones) crea un conjunto a partir de la lista de inscripciones.

6. Muestra el conjunto.

7. ¿Cuántos participantes diferentes hay?.

8. Añade a Elena al conjunto.

9. Comprueba si Marta está inscrita usando in.

10. Comprueba si Sonia NO está inscrita, usa el not.

```python
# 1. Muestra la lista.
print(inscripciones)
```

```python
# 2. \¿Cuántas inscripciones hay?.
len(inscripciones)
```

```python
# 3. Añade a Carlos al final.
inscripciones.append("Carlos")
```

```python
# 4. Vuelve a mostrar la lista y el número de inscripciones.
```

```python
print(inscripciones)
len(inscripciones)
```

```python
# 5. Ahora convierte la lista en un conjunto. Pista: set(inscripciones) crea un conjunto a partir de la lista de inscripciones.
inscripciones_cto = set(inscripciones)
```

```python
# 6. Muestra el conjunto.
print(inscripciones_cto)
```

```python
# 7. ¿Cuántos participantes diferentes hay? 
len(inscripciones_cto)
```

```python
# 8. Añade a Elena al conjunto.
inscripciones_cto.add("Elena")
```

```python
# 9. Comprueba si Marta está inscrita usando in.
"Marta" in inscripciones_cto
```

```python
# 10. Comprueba si Sonia no está inscrita
not "Sonia" in inscripciones_cto
```
