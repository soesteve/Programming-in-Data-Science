# Funciones definidas

<hr>

Una __función__ es un bloque de código **reutilizable** que realiza una tarea específica. Permite evitar repetir código, organizar mejor los programas y hacerlos más fáciles de mantener.

## Definición y llamada de una función

Una función se define con la palabra reservada `def` y forma la *cabecera* de la función. Además, toda función tiene un *cuerpo* que son las instrucciones que aparecen debajo de la cabecera indentadas.

La recomendación oficial de Python para la indentación es utiliza 4 espacios por cada nivel y no mezclar tabulaciones y espacios.


>```python
>def nombre_funcion(parametros):
>    instrucciones
>```

Un ejemplo de definición de función es:


````python
def area_rectangulo(base, altura):
    """
    Calcula el área de un rectángulo.
    """
    area = base * altura
    return area

# Llamadas a la función
area_casa = area_rectangulo(8, 9)
area_despacho = area_rectangulo(3, 5)
print(area_casa)
print(area_despacho)
````

## Partes de una función

### Cabecera:

- **def** indica que vamos a crear una función.

- **area_rectangulo** es el nombre de la función.

- **base** y **altura** son los parámetros, es decir, los datos que recibe la función.

### Cuerpo

- Una o más instrucciones (`area = base * altura`)

- Resultado que se devuelva para que pueda utilizarse fuera de la función (`return area`) 

El cuerpo de la función se compone de instrucciones, que trabajan con los parámetros.
Además hay una instrucción especial, llamada *return* sirve para expresar el valor que la función devuelve, tras los cálculos. 

El cuerpo de la función está indentado por unos espacios en blanco que están en el lado izquierdo de cada línea de código. En Python se usa la indentación para definir bloques de código. Las líneas que estén indentadas al mismo nivel pertenecen al mismo bloque de código.

Cada llamada de la función se hace con unos parámetros concretos, es decir, con los datos que realmente queremos hacer los cálculos. En cada llamada, los parámetros de entrada asumen los valores concretos y se ejecuta el cuerpo dichos datos. 

# Documentación de una función (docstring)

Debemos indicar lo que hace la función

````python
def area_rectangulo(base, altura):
    """
    Calcula el área de un rectángulo.
    """
    area = base * altura
    return area
````

# Anotaciones de tipos

Aunque la *anotación de tipos* no es obligatoria en Python, nos ayuda a entender las definiciones que escribimos hace tiempo o que han creado otras personas.

Las anotaciones de tipos no cambian el funcionamiento del programa, solo sirven para indicar qué tipo de datos reciben los parámetros de una función y qué tipo de dato devuelve.

```
def area_rectangulo(base: float, altura: float) -> float:
    """
    Calcula el área de un rectángulo.
    """
    area = base * altura
    return area
```


````python
print(area_rectangulo(8, 9))
print(area_rectangulo(3, 5))
````
