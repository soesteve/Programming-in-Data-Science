# Bases de Datos y Python

***Python permite trabajar con sistemas gestores de bases de datos como SQLite, MySQL o PostgreSQL***. Es decir, desde un programa podemos ejecutar instrucciones SQL para almacenar, consultar, modificar y eliminar información.

En estos apuntes, vamos a conectarnos a una base de datos ***SQLite*** desde Python y realizar algunas operaciones SQL como crear tablas, insertar, consultar, actualizar y eliminar registros.

Utilizamos SQLite porque:

- No necesita un servidor independiente.
- Guarda los datos en un archivo.
- Permite trabajar con SQL.
- Se integra con Python mediante `sqlite3`.




***Objetivos***

- Conectarse a una base de datos.
- Crear tablas e insertar registros con  `CREATE TABLE` y `INSERT INTO`.
- Ejecutar consultas `SELECT`.
- Actualizar o modificar registros con `UPDATE`. 
- Recuperar resultados con `fetchone()` y `fetchall()`.
- Eliminar registros usando `DELETE`.



***Python y SQLite no compiten: se complementan.***

- Python es un lenguaje de programación general que permite desarrollar algoritmos y aplicaciones.
- SQLite es un sistema gestor de bases de datos que permite almacenar, organizar, relacionar y consultar información.
- SQL es el lenguaje que utilizamos para comunicarnos con la base de datos.


Las base de datos trabajan con tablas, registros y campos mientras que los ficheros cvs son filas y columnas. Además, en las base de datos se pueden definir claves, restricciones, reglas de integridad y establecer relaciones entre tablas. Otra cualidad es que los SGBD ofrecen mecanismos para gestionar el acceso concurrente.



## Importar el módulo `sqlite3` y conectarse `connect`

La instrucción `sqlite3.connect()` establece una conexión. Si el archivo no existe, SQLite lo crea. En estos ejemplos utilizaremos `./BBDD/tienda.db`, que se guardará en el subdirectorio `./BBDD`.


```python
import sqlite3

conexion = sqlite3.connect("./BBDD/tienda.db")
print("Conexión establecida: ", conexion)
```


## Crear una tabla `CREATE`

Un *cursor* es un objeto que permite ejecutar instrucciones SQL sobre una base de datos y recuperar los resultados de las consultas. Un cursor permite ejecutar sentencias SQL. Hacemos un `commit()` para confirmar los cambios y `close()` cierra la conexión.

Vamos a crear una tabla `productos` que tendrá cuatro columnas: `id` (identificador), `nombre`, `precio` y `stock`.


```python

cursor = conexion.cursor()  # cursor permite ejecutar instrucciones SQL sobre la base de datos.

 # execute() ejecuta una sentencia SQL.
cursor.execute("""       
    CREATE TABLE IF NOT EXISTS productos (
        id INTEGER PRIMARY KEY,
        nombre TEXT NOT NULL,
        precio REAL,
        stock INTEGER
    )
""")

conexion.commit()    # Guarda los cambios realizados en la base de datos.
conexion.close()     # Cierra la conexión con la base de datos.

print("Tabla productos creada (si no existía).")
```

En este ejemplo hemos creado una tabla llamada `productos` con cuatro columnas:
- id (INTEGER): Identificador del producto.
- nombre (TEXT): Nombre del producto.
- precio (REAL): Precio del producto.
- stock	(INTEGER): Unidades disponibles.

## Insertar datos: `INSERT INTO`

***Insertar un registro***

La instrucción `INSERT INTO` añade un registro a una tabla.
En concreto, este programa inserta un producto llamado `Teclado`, con un `precio` de $25,50$ y un `stock` de $10$ unidades. Como la columna `id` es una clave primaria entera y no se especifica su valor, SQLite asignará $1$ automáticamente en este caso.

```python
conexion = sqlite3.connect("./BBDD/tienda.db")
cursor = conexion.cursor()

nombre = "Ratón"
precio = 15.90
stock = 20

cursor.execute("""
    INSERT INTO productos (nombre, precio, stock)
    VALUES (?, ?, ?)
""", (nombre, precio, stock))

conexion.commit()

```


**Atención:** si ejecutas varias veces la celda anterior, se insertará el mismo producto repetidamente.


## Consultar datos: `SELECT`  

La instrucción `SELECT` recupera información. Para ejecutar la consulta usamos `execute()` y para recoger resultados podemos usar `fetchone()`, `fetchall()` o `fetchmany()`.

***Consultar todos los registros***


```python

cursor.execute("SELECT * FROM productos")
productos = cursor.fetchall()

for producto in productos:
    print(producto)
```

Cada registro se devuelve como una tupla de Python, por ejemplo: `(1, 'Teclado', 25.5, 10)`.

***Consultar determinadas columnas***

Podemos indicar solo las columnas que queremos recuperar.


```python
cursor.execute("SELECT nombre, precio FROM productos")
productos = cursor.fetchall()

for producto in productos:
    print(producto)
```

***Consultar con una condición: `WHERE`***

La cláusula `WHERE` selecciona los registros que cumplen una condición. En este ejemplo, consultamos productos con precio inferior a 20 euros.


```python
cursor.execute("""
    SELECT nombre, precio
    FROM productos
    WHERE precio < ?
""", (20,))

for producto in cursor.fetchall():
    print(producto)
```

## Modificar con `UPDATE`

***Actualizar con `UPDATE`***

`UPDATE` modifica los registros que cumplen la condición `WHERE`. Si se omite `WHERE`, se modificarán todos los registros de la tabla.

Por ejemplo, vamos a modificar registros de la tabla productos con id igual a 10 y ponemos su precio  110.


```python

cursor.execute("""
    UPDATE productos
    SET precio = ?
    WHERE id = ?
""", (100, 3))

conexion.commit()
```

```python
cursor.execute("SELECT * FROM productos")
productos = cursor.fetchall()

for producto in productos:
    print(producto)
```

## Eliminar con `DELETE`

`DELETE` elimina los registros que cumplen la condición. Si se omite `WHERE`, se eliminarán todos los registros de la tabla. Comprueba siempre la condición antes de ejecutar la consulta.


```python

cursor.execute("""
    DELETE FROM productos
""")

conexion.commit()
conexion.close()
```
