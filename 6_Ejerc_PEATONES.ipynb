### Fechas 
En Python podemos trabajar con distintos tipos de datos temporales:
- `date`: representa una fecha: año, mes y día.
- `time`: representa una hora: hora, minutos, segundos.
- `datetime`: representa fecha y hora conjuntamente

Como hemos visto, pandas todavía no reconoce `FECHA` como una fecha, sino que la ha leído como texto. Por lo tanto usamos `pd.to_datetime()` para convertir los tipos de los valores. Además, necesitamos el parámetro `dayfirst = True` para indicar a pandas que, al interpretar una fecha, el primer número corresponde al día pues la fecha está escrita con el día antes del mes.

Ejecuto `df["FECHA"] = pd.to_datetime(df["FECHA"], dayfirst=True)` pero me da error.

````python
# df["FECHA"] = pd.to_datetime(df["FECHA"], dayfirst=True)
````

Y me da error, pero... ¿qué ha pasado?

Pues que en el fichero hay dos formatos diferentes en las fechas, unos es con la hora `01/01/2020 00:00` y el otro es sin hora `02/09/2020`.
Pandas ha inferido inicialmente el formato %d/%m/%Y %H:%M pero ha detectado un valor que no seguía este formato.
Si miramos al final del error, 

>ValueError: time data "02/09/2020" doesn't match format "%d/%m/%Y %H:%M", at position 5880)

nos dice que en la posición 5880 se ha encontrado 02/09/2020 sin hora 
y nos aconseja hacer varias cosas, la más interesante es la que dice: 

>`format='mixed'`, and the format will be inferred for each element individually. You might want to use `dayfirst` alongside this.


Vuelve a realizar la conversión de FECHA, pero esta vez teniendo en cuenta el mensaje de error.

````python
df["FECHA"] = pd.to_datetime(
    df["FECHA"],
    dayfirst = True,
    format = "mixed"
)
````

## Total de peatones de Madrid por mes

## Calcular el total de peatones en una fecha concreta y por distrito

Queremos saber cuántos peatones había en la calle, por ejemplo, el 15 de marzo del 2020. Para esto hay que seguir los siguientes pasos:

1. A partir del DataFrame df_1, seleccionar los registros del 15 de marzo de 2020 y dejarlos en un nuevo 'df_fecha`.

2. Calcular el total de peatones aplicando el método `sum()` a la variable `PEATONES`.

3. Calcular el total por distrito

````python
# 1. Seleccionar los registros del 15 de marzo de 2020

fecha = pd.to_datetime("15/03/2020", dayfirst=True)

df_fecha = df_1[df_1["FECHA"].dt.date == fecha.date()].copy()

# 2. Calcular el total de peatones

total_peatones = df_fecha["PEATONES"].sum()

print("Total de peatones:", total_peatones)

# 3. Calcular el total por distrito

peatones_distrito = df_fecha.groupby("DISTRITO")["PEATONES"].sum()

peatones_distrito = peatones_distrito.sort_values(ascending=False)

print(peatones_distrito)
````

## Las 10 calles con más peatones

Pasos:

1. A partir del DataFrame df_1, agrupa los registros por calle, utilizando la columna DIRECCION.

2. Calcula el total de peatones registrados en cada calle.

3. Ordena las calles de mayor a menor número de peatones.

4. Muestra las 10 calles con mayor número total de peatones.


````python
# 1. Calcular el total de peatones por calle

peatones_calle = df_1.groupby("DIRECCION")["PEATONES"].sum()

# 2. Ordenar de mayor a menor

peatones_calle = peatones_calle.sort_values(ascending=False)

# 3. Seleccionar las 10 calles con más peatones

top_10 = peatones_calle.head(10)

print(top_10)
````

## ¿A qué hora hay más peatones?

Pasos:

1. Agrupa los registros por hora.

2. Calcula el total de peatones registrados en cada hora.

3. Ordena los resultados de mayor a menor.

4. Identifica la hora con mayor y menor número total de peatones.



````python
# 1. Crear una columna con la hora
# Como FECHA contiene la fecha y la hora, podemos extraer la hora con .dt.hour
df_1["HORA_DIA"] = df_1["FECHA"].dt.hour

# 2. Calcular el total de peatones por hora
peatones_hora = df_1.groupby("HORA_DIA")["PEATONES"].sum()

# 3. Ordenar de mayor a menor
peatones_hora_ordenado = peatones_hora.sort_values(
    ascending=False
)

print(peatones_hora_ordenado)
````
