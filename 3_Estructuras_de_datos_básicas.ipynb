# Estructuras de datos básicas

Las estructuras de datos permiten almacenar y organizar información. Python incluye cuatro estructuras fundamentales:

- Listas (list)

- Tuplas (tuple)

- Conjuntos (set)


## Listas (list)

Las listas son colecciones ordenadas y modificables. Pueden contener elementos de distintos tipos (característica no común con otros lenguajes de programación).


    

```python
# Crear una lista
frutas = ["manzana", "pera", "plátano"]

print(frutas)
```

### Acceso a los elementos de una lista

Se acceden a los elementos mediante el operador `[ ]`. Los elementos de una lista están indexados y los índices comienzan en 0.

```python
print(frutas[0])      # Primer elemento
print(frutas[-1])     # Último elemento
```

### Modificar una lista
Los elementos de la lista se modifican usando la asignación (=).

```python
frutas[1] = "uva"
print(frutas)
```

### Operaciones frecuentes

Las operaciones definidas sobre listas las podemos encontrar en la documentación de Python (https://docs.python.org/es/3/tutorial/datastructures.html).
No obstante, a continuación podemos ver una muestra:

```python
frutas.append("kiwi")      # Añadir al final
print(frutas)
frutas.insert(1, "fresa")  # Insertar en una posición
print(frutas)
frutas.remove("uva")       # Eliminar por valor
print(frutas)
frutas.pop()      # Eliminar el último
print(frutas)
frutas.sort()      # Ordenación lexicográfica
print(frutas)

len(frutas)     # Tamaño, o longitud, de la lista 
```

## Tuplas (tuple)

Las tuplas son objetos inmutables pero con orden. Una vez creadas, no pueden modificarse.
Se suelen usar para coordenadas, fechas fijas...

```python
coordenada = (10, 20)
print(coordenada)
```

### Acceso

```python
print(coordenada[0])
```

### Desempaquetado

```python
x, y = coordenada
print(x, y)
```

## Conjuntos (set)

Los conjuntos almacenan elementos sin orden y sin duplicados (https://docs.python.org/es/3/tutorial/datastructures.html#sets).
Algunas operaciones sobre conjuntos son:

```python
numeros = {1, 2, 3, 3, 4}  # Def un cto con {}
print(numeros)
numeros.add(5)     # Añade
numeros.add(3)     # Añade
print(numeros)
numeros.remove(2)  # Elimina
print(numeros)

print('---')

A = {2,4,6}
B = {3,4,5}

print(A | B)   # Unión
print(A & B)   # Intersección
print(A - B)   # Diferencia
print(4 in A)      # pertenencia, comprueba si un elemento pertenece a un conjunto
print(44 in A)
```

# Comparación de las estructuras

|Estructura|Ordenada|Modificable|Duplicados|
|----------|:--------:|:-----------:|:----------:|
|Lista|Sí|Sí|Sí|
|Tupla|Sí|No|Sí|
|Conjunto|No|Sí|No|


# Transformaciones

Una estructura de datos se puede transformar en otra 

```python
A = [3, 5, 1, 4, 4, 3, 2, 1]  # Lista [] con corchetes 
tuple_A = tuple(A)
print(tuple_A, "   Obs: tupla ( ) con paréntesis")
set_A = set(A)
print(set_A, "   Obs: cto { } con llaves")

B = {2, 5, 6, 1}  # cto { } con llaves
list_B = list(B)
print(list_B, "   Obs:  Lista [] con corchetes") 
```
