# Elementos básicos del lenguaje

En esta Sección se introducen los conceptos de variable y asignación del lenguaje Python.

## Variable

Cuando escribimos un programa necesitamos guardar información para poder utilizarla más adelante. En Python esa información se almacena en variables.

Una __variable__ es un nombre que hace referencia a un valor almacenado en la memoria del programa. Podemos imaginar una variable como una etiqueta pegada a una caja: la etiqueta es el nombre de la variable y dentro de la caja está el valor.

 En Python, una variable se define con la sintaxis:

```
nombre_de_la_variable = valor_de_la_variable
```

Por ejemplo:

````python
edad = 25
````

Aquí:

- edad es el nombre de la variable.

- 25 es el valor almacenado.

Podemos mostrar su contenido con la función predefinada de python: print().

````python
print(edad)      # print es una función predefinida de python
````

Los nombres de las variables en Python pueden contener caracteres alfanuméricos a-z, A-Z, 0-9 y otros símbolos como la \_. 

Por cuestiones de estilo, las variables suelen empezar con minúscula. Podéis consultar la guia de estilo para código escrito en Python en el siguiente enlace [https://www.python.org/dev/peps/pep-0008/](https://www.python.org/dev/peps/pep-0008/)

> Algunas palabras no pueden ser usados como nombres de variables porque son palabras reservadas
> 
>    and, as, assert, break, class, continue, def, del, elif, else, except, exec, finally, for, from, global, if, import, in, is, lambda, not, or, pass, print, raise, return, try, while, with, yield

## Asignación

La __asignación__ es la operación que guarda un valor en una variable mediante el símbolo __=__. 
Este signo no significa igualdad matemática, lo que hace es asignar el valor de la derecha a la variable de la izquierda. En la izquierda siempre hay una variable.

### Nuevo valor      

Una variable puede recibir un nuevo valor en cualquier momento.  

````python
asignatura = "Programación"
print(asignatura)
asignatura = "Programación de Ciencia de Datos"
print(asignatura)
````

### Asignación utilizando otra variable

También podemos copiar el valor de una variable a otra.

````python
a = 10
b = a

print(a)
print(b)
````

Ambas variables contienen inicialmente el mismo valor.

###  Actualizar una variable

Una operación muy frecuente consiste en actualizar una variable utilizando su valor anterior. Es decir, el valor de la variable, puede  ```variar```. 



Por ejemplo, podemos sumar 1 a la edad, cuando cumplimos años :-)


````python
edad = 25
edad = edad + 1.  # existe la forma abreviada edad += 1
print(edad)
````

#### Formateando cadenas

El uso del símbolo __`%`__ seguido de uno o mas caracteres de formato permiten controlar con precisión la construcción de  cadenas. Por ejemplo, creamos una cadena:

````python
texto = "Cambio: %d %s son %.2f %s"
texto % (1, "dolar" , 0.926666666 , "euros")
````

>  __Operadores de formato__

> `%s` : ha de sustituirse por una cadena

>  `%d` : ha de sustituirse por un entero

>  `%.2f` : ha de sustituirse por un número con dos decimales

````python
radio = 4.323343
pi = 3.1416
area = pi*radio**2

print ('El área de un círculo de radio %.3f es %.5f' % (radio, area))
````

````python
print ('El área de un círculo de radio %.3f es %.5f' % (radio, pi*radio**2))
````

También podemos usar la función `format` del siguiente modo:

````python
'El área de un círculo de radio {:.3f} es {:.4f}'.format(radio, pi*radio**2)
````

## Curiosidades

````python
def paradoja(a):
    b = math.sqrt(a)         # la raíz cuadrada de 2
    return a == b * b        # esto debería ser cierto siempre, no lo es por el error de la representación de los reales

paradoja(2)
````

````python
def fibonacci(n):
    phi = (1 + math.sqrt(5)) / 2
    return (phi ** n - (1 - phi) ** n)/math.sqrt(5)

fibonacci(4)
````

````python
def fibonacci(n):
    phi = (1 + math.sqrt(5)) / 2
    return int(round((phi ** n - (1 - phi) ** n)/math.sqrt(5)))

fibonacci(4)
````
