# Elementos básicos del lenguaje

En esta Sección se introducen los conceptos de variable y asignación del lenguaje Python.

## Variable

Cuando escribimos un programa necesitamos guardar información para poder utilizarla más adelante. En Python esa información se almacena en variables.

Una __variable__ es un nombre que hace referencia a un valor almacenado en la memoria del programa. Podemos imaginar una variable como una etiqueta pegada a una caja: la etiqueta es el nombre de la variable y dentro de la caja está el valor.

 En Python, una variable se define con la sintaxis:

```
nombre_de_la_variable = valor_de_la_variable
```

Por ejemplo:

````python
edad = 25
````

Aquí:

- edad es el nombre de la variable.

- 25 es el valor almacenado.

Podemos mostrar su contenido con la función predefinada de python: print().

````python
print(edad)      # print es una función predefinida de python
````

Los nombres de las variables en Python pueden contener caracteres alfanuméricos a-z, A-Z, 0-9 y otros símbolos como la \_. 

Por cuestiones de estilo, las variables suelen empezar con minúscula. Podéis consultar la guia de estilo para código escrito en Python en el siguiente enlace [https://www.python.org/dev/peps/pep-0008/](https://www.python.org/dev/peps/pep-0008/)

> Algunas palabras no pueden ser usados como nombres de variables porque son palabras reservadas
> 
>    and, as, assert, break, class, continue, def, del, elif, else, except, exec, finally, for, from, global, if, import, in, is, lambda, not, or, pass, print, raise, return, try, while, with, yield

## Asignación

La __asignación__ es la operación que guarda un valor en una variable mediante el símbolo __=__. 
Este signo no significa igualdad matemática, lo que hace es asignar el valor de la derecha a la variable de la izquierda. En la izquierda siempre hay una variable.

### Nuevo valor      

Una variable puede recibir un nuevo valor en cualquier momento.  

````python
asignatura = "Programación"
print(asignatura)
asignatura = "Programación de Ciencia de Datos"
print(asignatura)
````

### Asignación utilizando otra variable

También podemos copiar el valor de una variable a otra.

````python
a = 10
b = a

print(a)
print(b)
````

Ambas variables contienen inicialmente el mismo valor.

###  Actualizar una variable

Una operación muy frecuente consiste en actualizar una variable utilizando su valor anterior. Es decir, el valor de la variable, puede  ```variar```. 



Por ejemplo, podemos sumar 1 a la edad, cuando cumplimos años :-)


````python
edad = 25
edad = edad + 1.  # existe la forma abreviada edad += 1
print(edad)
````

## Los tipos de datos en  Python 



Python es un lenguaje de tipado dinámico, es decir, el tipo se determina en el momento de la asignación.  

Cada variable, tiene un nombre y un valor. El valor define el tipo de datos de la variable. 

###  Tipo `int`: números enteros

Se trata de números sin decimales.

````python
entero = 10
````

Para conocer el tipo de una variable podemos ejecutar la función ```type``` de Python.

````python
type(entero)
````

###  Tipo `float`: números reales

Los valores reales se representan mediante el punto, separando así la parte entera de la parte decimal. Estos valores decimos que son de tipo `float`.

````python
decimal = 3.14
````

````python
type(decimal)
````

Es posible mezclar en una misma expresión datos de tipo entero y real. Si alguno de los operandos es real, el resultado es también real:

````python
precio = 6
print(precio / 2)
type(precio / 2)
````

````python
print(precio * 2)
type(precio * 2)
````

### Funciones predefinidad sobre los enteros y reales

Python proporciona funciones  que podemos utilizar en las expresiones. Estas funciones se dice que están __predefinidas__ (como la función `print`):

````python
# Valor absoluto
abs( 2 - 8 )
````

````python
max(5 , 8, 10 , 7)   # máximo de un conjuto de valores (separados por comas)
````

````python
min( -1 , 1 , 0, -3, 3 )    # mínimo de un conjuto de valores
````

````python
round( 24.7555, 2 )       # redondea un float a 2 decimales
````

````python
round( 18.6555 )       # redondea un float a 2 decimales
````

Podemos __convertir expresiones__ a `int, float, bool, str`  mediante funciones:

````python
int( 18.6 )
````

````python
float( 1 )
````

````python
float( 8 + 9.9 )
````

````python
complex(2)
````

````python
str ( 256568 )    # convierte un entero a una cadena de caracteres
````

````python
float('67.8888')
````

````python
edad = 120
````

````python
'Tengo ' + str(edad) + ' años'
````

###  Tipo `bool`: datos Boolean

bool representa valores de verdadero o falso. Solo existen dos posibles valores: `True` y `False`. 

````python
v = True  # cierto
f = False  # falso
````

Los operadores booleanos son los habituales, `and` para la conjunción, `or` para la disyunción y  `not` para la negación.

````python
a = True
a and False
````

````python
a or False
````

````python
not a 
````

````python
print( a )
````

````python
a = not a
````

````python
print(a)
````

````python
x = 6
x < 5
````

Además dispone de los operadores habituales de comparación. Los operadores de comparación son:

* `==` igual a
* `!=` distinto de 
* `<` menor que
* `<=` menor o igual que
* `>` mayor que
* `>=` mayor o igual que

> Devolverán un valor booleano: `True` o `False`

````python
x = 2
6  ==  x 
````

````python
a = True
b = False
a == b      
  #  la comparación devuelve un valor lógico
````

````python
print( a, b )        # la función print con dos parámetros
                   # imprime el valor de a seguido el valor de b
````

````python
x = 6
y = 9
print('1.', x < y)      # la función print con dos parámetros:
                        #   el primero es una cadena de caracteres
                        #   el segundo es una expresión que se evaluará a True o False
print('2.' , x <= y)
print('3.' , x > y )
print('4.' , x >= y)
````

````python
'aaab' > 'ba'    # comparación de cadenas según el órden lexicográfico
````

### Tipo `str`: cadena de caracteres

`str` es el tipo de dato que representa una cadena de caracteres, es decir, texto.
Este tipo se corresponde con el tipo `String` de otros lenguajes. 

Las cadenas de caracteres se definen mediante un texto encerrado entre comillas dobles o simples. 

````python
objetivo = "Aprender a manejar Python"
meta = 'Ser toda una experta'
print(objetivo)
print(meta)
````

Las cadenas también se pueden comparar.

````python
a = 'Sonia'
b = "SONIA"
a == b
````

Las cadenas de caracteres se pueden concatenar con el operador `+`.

````python

b = objetivo + " y " + meta 
b
````

````python
num = 10
texto = 'Mostrar el número ' + str(num)
texto
````

Se puede utilizar los operadores de comparación teniendo en cuenta que el orden entre cadenas es el orden lexicográfico.

````python
"Juan" < "Juap"
````

````python
"j" == "J" # las mayúsculas y las minúsculas son diferentes
````

Para conocer el tamaño de una cadena, se utiliza la función `len` como se muestra a continuación:

````python
a
````

````python
len(a)
````

__Caracteres especiales o secuencias de escape__

````python
especial = "Las \"comillas\" son caracteres epeciales"
especial
````

````python
especial = "Hay otros caracteres especiales como el salto de línea (\\n).\nSirve para representar el fín de línea en ficheros."
print(especial)
````
