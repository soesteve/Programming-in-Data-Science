# Estructuras de datos básicas

Las estructuras de datos permiten almacenar y organizar información. Python incluye cuatro estructuras fundamentales:

- Listas (list)

- Tuplas (tuple)

- Conjuntos (set)

- Diccionarios (dict)



## Listas (list)

Las listas son colecciones ordenadas y modificables. Pueden contener elementos de distintos tipos (característica no común con otros lenguajes de programación).


    

````python
# Crear una lista
frutas = ["manzana", "pera", "plátano"]

print(frutas)
````

### Acceso a los elementos de una lista

Se acceden a los elementos mediante el operador `[ ]`. Los elementos de una lista están indexados y los índices comienzan en 0.

````python
print(frutas[0])      # Primer elemento
print(frutas[-1])     # Último elemento
````

### Modificar una lista
Los elementos de la lista se modifican usando la asignación (=).

````python
frutas[1] = "uva"
print(frutas)
````

### Operaciones frecuentes

````python
frutas.append("kiwi")      # Añadir al final
print(frutas)
frutas.insert(1, "fresa")  # Insertar en una posición
print(frutas)
frutas.remove("uva")       # Eliminar por valor
print(frutas)
frutas.pop()      # Eliminar el último
print(frutas)
````

````python
len(frutas)    # Longitud
````

## Tuplas (tuple)

Las tuplas son ordenadas pero inmutables. Una vez creadas, no pueden modificarse.
Se suelen usar para coordenadas, fechas fijas...

````python
coordenada = (10, 20)
print(coordenada)
````

### Acceso

````python
print(coordenada[0])
````

### Desempaquetado

````python
x, y = coordenada
print(x, y)
````

## Conjuntos (set)

Los conjuntos almacenan elementos sin orden y sin duplicados.

````python
numeros = {1, 2, 3, 3, 4}
print(numeros)
````

Las operaciones disponibles son:

````python
numeros.add(5)     # Añade
print(numeros)
numeros.remove(2)  # Elimina
print(numeros)

print('---')

A = {2,4,6}
B = {3,4,5}

print(A | B)   # Unión
print(A & B)   # Intersección
print(A - B)   # Diferencia
````

Para comprobar si un elemento pertenece a un conjunto, se utiliza el operador in, que devuelve un valor booleano (`True` o `False`).

La sintáxis es:
```python
         elemento in conjunto
```

````python
4 in A, 44 in A
````

## Diccionarios (dict)

Los diccionarios almacenan información mediante pares clave-valor. La clave es única.

````python
persona = {
    "nombre": "Sonia",
    "edad": 25,
    "ciudad": "Madrid"
}

print(persona)
````

Para acceder a los valores del diccionario usamos `dicc[calve]`

````python
print(persona["nombre"])
print(persona["edad"])
````

Podemos insertar/modificar los valores de un diccionario, las claves no.

````python
persona["edad"] = 26
persona["profesion"] = "Matemática"
print(persona)
````

Podemos eliminar una entrada o fila

````python
del persona['ciudad']
````

# Comparación de las estructuras

|Estructura|Ordenada|Modificable|Duplicados|
|----------|:--------:|:-----------:|:----------:|
|Lista|Sí|Sí|Sí|
|Tupla|Sí|No|Sí|
|Conjunto|No|Sí|No|
|Diccionario|Sí|Sí|Claves únicas|
