# Exploración y manipulación de Datos con Pandas

La manipulación de datos consiste en leer, organizar y modificar información almacenada en archivos o bases de datos. Python utiliza principalmente la biblioteca **Pandas** para estas tareas.

Para importar esta biblioteca escribimos

```python
import pandas as pd
```

La estructura principal de Pandas es el **DataFrame**, una tabla con filas y columnas similar a una hoja de cálculo (Excel, Number, Google sheet,...).
Podemos crear un DataFrame manualmente o importarlo desde un fichero.

## Crear un DataFrame manualmente:

````python
import pandas as pd

datos = {
    "Nombre": ["Ana", "Luis", "Marta"],
    "Edad": [24,31,28],
    "Ciudad": ["Madrid","Sevilla","Valencia"]
}

df = pd.DataFrame(datos)

print(df)
````

## Importar los datos desde un fichero


A la vez que importamos los datos, creamos el dataframe. Los datos pueden estar almacenados en distintos formatos. Los más habituales son:

***Leer un archivo CSV***

Un archivo CSV es un texto donde las columnas están separadas por comas.


````python
df = pd.read_csv("./Data/Nombre_Edad_Ciudad.csv")
df.head()    # Muestra las primeras filas
````

***Leer un archivo Excel***


````python
import openpyxl
df = pd.read_excel("./Data/Nombre_Edad_Ciudad.xlsx")
df.head()    # Muestra las primeras filas
````

## Explorar filas y columnas

Una tabla puede tener un gran número de entradas o filas, es por ello que se recomienda explorar una parte del dataframe con `head` y `tail`.

````python
df.head()  # Primeras filas
````

````python
df.tail()  # Últimas filas
````

Puede interesarnos una  varias columnas. Podemos seleccionar el número de columnas que queramos mediante los corchetes.

````python
df["Nombre"]
````

````python
df[["Nombre","Edad"]]
````

También podemos seleccionar unas filas concretas

````python
df.iloc[0:2]
````

## Filtrar

Se **filtran** las filas u observaciones que cumplen una determinada restricción.

````python
df[df["Edad"]>=30]
````

````python
df[(df["Edad"]>=20) & (df["Edad"]< 40)]
````

## Ordenar

Se pueden **ordenar** los datos por determinadas columnas.

````python
df.sort_values("Edad")
````

````python
df.sort_values("Edad", ascending=False)
````

## Crear nuevas columnas

Podemos crear nuevas columnas resultado de
 realizar cálculos.

````python
df["Menor_edad"] = df["Edad"]>=18
df
````

## Eliminar columnas

````python
df = df.drop(columns="Menor_edad")
df
````

## Renombrar columnas

````python
df = df.rename(columns={"Ciudad":"Población"})
df
````

## Valores nulos

````python
df.isnull()
````

## Eliminar filas con datos vacíos:

````python
df.dropna()
````

## Operaciones sobre una columna numérica

````python
df["Edad"].mean()
````

````python
df["Edad"].max()
````

````python
df["Edad"].min()
````

````python
df["Edad"].sum()
````

## Guardar datos

````python
# Guardar como CSV:

df.to_csv("./Data/nuevo.csv", index=False)

# Guardar como Excel:

df.to_excel("./Data/nuevo.xlsx", index=False)
````
